## Qualitative samples per model

```python
for model in ('constant', 'retrieval', 'group_template'):
    df = pd.read_parquet(config.GENERATIONS / f'{model}_test_scored.parquet')
    print(f'\n=== {model} (test) ===')
    for _, row in df.sort_values('rougeL', ascending=False).head(2).iterrows():
        print('TICKET:', row['ticket'][:200], '...')
        print('GOLD  :', row['gold'][:200], '...')
        print('PRED  :', row['pred'][:200], '...')
        print('rougeL=', round(row['rougeL'], 3), 'tfidf_cos=', round(row['tfidf_cos'], 3))
        print('-'*40)
```
